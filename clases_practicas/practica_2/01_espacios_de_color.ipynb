### Ejercicio: Separar la mesa (color verde) del resto de la escena

![image](./imgBloque1.bmp)

```python
# Primera parte: encontrar un espacio de color conveniente para la discriminación

img_color = cv.imread('imgBloque1.bmp')

# Convertir la imagen a HSV
imgHSV = cv.cvtColor(img_color, cv.COLOR_BGR2HSV)
plt.figure()
plt.imshow(imgHSV)
plt.title('Imagen en espacio de color HSV')
plt.show()
```

```python
im_h, im_s, im_v = cv.split(imgHSV)

# La curva de Hue, discrimina fácilmente el color del tablero de los colores del
# fondo

plt.figure()
plt.imshow(im_h, cmap='jet')
plt.title('Hue')
plt.show()

# Puedo complementar con la curva de saturación, ya que ambas son muy uniformes 
# sobre el tablero

plt.figure()
plt.imshow(im_s,cmap='jet')
plt.title('Saturation')
plt.show()

# La imagen de intensidad (V) permite discriminar fácilmente el bloque del 
# tablero, pero confunde el tablero con el fondo

plt.figure()
plt.imshow(im_v,cmap='gray')
plt.title('Value')
plt.show()
```

```python
# Hacemos una descripcion estadistica
muestra = imgHSV[300:350, 300:350, :]

color_h_medio, _, _ = np.mean(muestra, axis=(0,1))
print(f'Valor medio de la componente Hue: {color_h_medio}')

muestra = cv.cvtColor(muestra, cv.COLOR_HSV2BGR)

plt.figure()
plt.imshow(muestra)
plt.title('Muestra de la mesa')
plt.show()
```

```python
SPAN_RANGE = 10
color_l = (color_h_medio - SPAN_RANGE, 0, 0)
color_u = (color_h_medio + SPAN_RANGE, 180, 180)

mask = cv.inRange(src=imgHSV, lowerb=color_l, upperb=color_u)

plt.imshow(mask, cmap='gray')
plt.title('Mascara de segmentacion')
plt.show()
```

```python
img_segmentada = cv.bitwise_and(img_color, img_color, mask=mask)
plt.imshow(img_segmentada[:, :, ::-1]) # bgr to rgb
plt.title('Objeto segmentado')
plt.show()
```
